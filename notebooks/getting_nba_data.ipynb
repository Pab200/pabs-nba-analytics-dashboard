```python
from nba_api.stats.static import players
import pandas as pd

df_players = pd.DataFrame(players.get_players())
df_players.to_csv("../data/raw/players.csv", index=False)
df_players.head()
```

```python
from nba_api.stats.static import teams

df_teams = pd.DataFrame(teams.get_teams())
df_teams.to_csv("../data/raw/teams.csv", index=False)
df_teams.head()
```

```python
from nba_api.stats.endpoints import leaguedashplayerstats

stats = leaguedashplayerstats.LeagueDashPlayerStats(
    season="2025-26",
    season_type_all_star="Regular Season",
    per_mode_detailed="PerGame"
)

df_stats = stats.get_data_frames()[0]
df_stats.to_csv("../data/raw/season_stats_2025-26.csv")
df_stats.head()
```

```python
from nba_api.stats.endpoints import drafthistory

# Pull full draft history
dh = drafthistory.DraftHistory(league_id='00')
df_draft = dh.get_data_frames()[0]

df_draft.to_csv("../data/raw/rookies.csv", index=False)
df_draft.head(10)
```
