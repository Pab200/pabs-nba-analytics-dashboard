```python
import pandas as pd
from IPython.display import display

df_players = pd.read_csv("../data/raw/players.csv")
df_teams = pd.read_csv("../data/raw/teams.csv")
df_stats = pd.read_csv("../data/raw/season_stats_2025-26.csv")
df_rookies = pd.read_csv("../data/raw/rookies.csv")

display(df_players.head())
display(df_teams.head())
display(df_stats.head())
display(df_rookies.head())

display(df_players.shape)
display(df_teams.shape)
display(df_stats.shape)
display(df_rookies.shape)

display(df_stats.columns)

display(df_stats.describe())

display(df_stats[["PLAYER_NAME", "PTS", "AST", "REB", "FG_PCT", "FG3_PCT"]].head())
```

```python
display(df_players['is_active'].value_counts())

active_players = df_players[df_players['is_active'] == True]
display(active_players.head())

display(len(active_players))

display(df_players[['full_name', 'id']].head(20))

display(df_players['full_name'].value_counts().head())

display(df_stats[['PLAYER_NAME', 'TEAM_ID', 'TEAM_ABBREVIATION']].head())
```

```python
display(df_teams.head())
display(df_teams.columns)

display(len(df_teams))

display(df_teams[['city', 'nickname', 'abbreviation']].head())

display(df_teams.sort_values('city').head(10))

display(df_teams[['id', 'full_name']].head())
```

```python
df_stats['TOTAL_3PA'] = df_stats['FG3A'] * df_stats['GP']
df_active = df_stats[df_stats['GP'] > 0]
team_3pa = df_active.groupby('TEAM_ABBREVIATION')['TOTAL_3PA'].sum().reset_index()
display(team_3pa.sort_values('TOTAL_3PA', ascending=False).head(10))

df_stats['TOTAL_MIN'] = df_stats['MIN'] * df_stats['GP']
display(df_stats[['PLAYER_NAME', 'MIN', 'GP', 'TOTAL_MIN']].sort_values('TOTAL_MIN', ascending=False).head(10))
```

```python
df_rookies_2025 = df_rookies[df_rookies['SEASON'] == 2025]

df_rookie_stats = df_stats.merge(
    df_rookies_2025[['PERSON_ID', 'PLAYER_NAME']],
    left_on='PLAYER_ID',
    right_on='PERSON_ID',
    how='inner'
)

df_rookie_stats[['PLAYER_NAME_x', 'REB', 'GP']].sort_values('REB', ascending=False).head(10)
```

```python
df_stats['PTS_TOTAL'] = df_stats['PTS'] * df_stats['GP']
df_stats['FGA_TOTAL'] = df_stats['FGA'] * df_stats['GP']
df_stats['FTA_TOTAL'] = df_stats['FTA'] * df_stats['GP']
df_stats['TOV_TOTAL'] = df_stats['TOV'] * df_stats['GP']
df_stats['OREB_TOTAL'] = df_stats['OREB'] * df_stats['GP']

df_stats['POSS'] = df_stats['FGA_TOTAL'] + 0.44 * df_stats['FTA_TOTAL'] + df_stats['TOV_TOTAL'] - df_stats['OREB_TOTAL']

df_team = df_stats.groupby('TEAM_ABBREVIATION').agg({
    'PTS_TOTAL': 'sum',
    'POSS': 'sum'
}).reset_index()

df_team['ORTG'] = (df_team['PTS_TOTAL'] / df_team['POSS']) * 100

df_team.sort_values('ORTG', ascending=False).head(10)
```

```python
display(df_stats.head())
display(df_stats.columns)
display(df_stats.describe())

display(df_stats[['PLAYER_NAME', 'PTS']].sort_values('PTS', ascending=False).head(10))

display(df_stats[['PLAYER_NAME', 'REB']].sort_values('REB', ascending=False).head(10))

display(df_stats[['PLAYER_NAME', 'AST']].sort_values('AST', ascending=False).head(10))

df_filtered = df_stats.copy()
df_filtered = df_filtered[df_filtered['FGA'] >= 10]
display(df_filtered[['PLAYER_NAME', 'FGA', 'FG_PCT']].sort_values('FG_PCT', ascending=False).head(10))


display(df_stats[['PLAYER_NAME', 'MIN', 'GP']].sort_values('MIN', ascending=False).head(10))
```
