```python
import pandas as pd
import matplotlib.pyplot as plt

from src.colors import get_primary, get_secondary
from src.metrics import add_advanced_metrics

plt.style.use("ggplot")
```

```python
players = pd.read_csv("./data/raw/players.csv")
teams = pd.read_csv("../data/raw/teams.csv")
stats = pd.read_csv("../data/raw/season_stats.csv")

players.head(), teams.head(), stats.head()
```

```python
print("Players:", players.shape)
print("Teams:", teams.shape)
print("Stats:", stats.shape)

players.info()
teams.info()
stats.info()
```

```python
players["full_name"] = players["full_name"].astype(str)
players["is_active"] = players["is_active"].astype(bool)

active_players = players[players["is_active"] == True]
inactive_players = players[players["is_active"] == False]

active_players.head()
```

```python
players["team_id"].value_counts().head()
players["full_name"].value_counts().head()
```

```python
teams["full_name"] = teams["full_name"].astype(str)
teams["abbreviation"] = teams["abbreviation"].astype(str)

teams.sort_values("city").head()
```

```python
teams[["id", "city", "nickname", "abbreviation"]].head(10)
```

```python
stats["PLAYER_NAME"] = stats["PLAYER_NAME"].astype(str)
stats["TEAM_ABBREVIATION"] = stats["TEAM_ABBREVIATION"].astype(str)
stats["SEASON"] = stats["SEASON"].astype(int)

stats.describe()
```

```python
seasons = sorted(stats["SEASON"].unique())
selected_season = seasons[-1]  # most recent
selected_season
```

```python
season_df = stats[stats["SEASON"] == selected_season].copy()
season_df.head()
```

```python
top_scorers = season_df.sort_values("PTS", ascending=False).head(10)
top_scorers
```

```python
colors = [get_primary(t) for t in top_scorers["TEAM_ABBREVIATION"]]
edges = [get_secondary(t) for t in top_scorers["TEAM_ABBREVIATION"]]

plt.figure(figsize=(10,6))
plt.bar(top_scorers["PLAYER_NAME"], top_scorers["PTS"],
        color=colors, edgecolor=edges, linewidth=3)
plt.xticks(rotation=45)
plt.title(f"Top Scorers — {selected_season}")
plt.show()
```

```python
season_adv = add_advanced_metrics(season_df)
season_adv.head()
```

```python
ts_leaders = season_adv.sort_values("TS_PCT", ascending=False).head(10)
ts_leaders
```

```python
colors = [get_primary(t) for t in ts_leaders["TEAM_ABBREVIATION"]]
edges = [get_secondary(t) for t in ts_leaders["TEAM_ABBREVIATION"]]

plt.figure(figsize=(10,6))
plt.bar(ts_leaders["PLAYER_NAME"], ts_leaders["TS_PCT"],
        color=colors, edgecolor=edges, linewidth=3)
plt.xticks(rotation=45)
plt.title(f"Top TS% — {selected_season}")
plt.show()
```

```python
team_3pa = (
    season_df.groupby("TEAM_ABBREVIATION")["FG3A"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

team_3pa.head()
```

```python
colors = [get_primary(t) for t in team_3pa["TEAM_ABBREVIATION"]]
edges = [get_secondary(t) for t in team_3pa["TEAM_ABBREVIATION"]]

plt.figure(figsize=(12,6))
plt.bar(team_3pa["TEAM_ABBREVIATION"], team_3pa["FG3A"],
        color=colors, edgecolor=edges, linewidth=3)
plt.xticks(rotation=45)
plt.title(f"Team 3PA — {selected_season}")
plt.show()
```
