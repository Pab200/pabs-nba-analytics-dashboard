```python
import os
import sys
import pandas as pd

# ------------------------------------------------------------
# Fix path: Navigate up 2 directories from notebooks/active/
# ------------------------------------------------------------
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), "..", ".."))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from src.fantasy.fetch_game_logs import fetch_all_players_game_logs

DATA_ROOT = os.path.join(PROJECT_ROOT, "data")
LOGS_ROOT = os.path.join(DATA_ROOT, "player_game_logs")

os.makedirs(LOGS_ROOT, exist_ok=True)

seasons = [
    "2020-21", "2021-22", "2022-23", "2023-24", "2024-25", 
    "2025-26",
]

for season in seasons:
    print(f"🔽 Downloading {season}...")

    season_folder = os.path.join(LOGS_ROOT, season)
    os.makedirs(season_folder, exist_ok=True)

    file_path = os.path.join(season_folder, f"player_game_logs{season}.csv")

    if os.path.exists(file_path):
        print(f"✔ Already exists - skipping {season}")
        continue

    try:
        df = fetch_all_players_game_logs(season)
        df.to_csv(file_path, index=False)
    except Exception as e:
        print(f"❌ Error downloading {season}: {e}")

print("🎉 All downloads complete!")
```
